```python
import sqlite3
```

```python

conn = sqlite3.connect('d:/AI_team/github/Vision_AI_RnD_team/projects/test_project/ais_weather/korea_weather.db', check_same_thread=False)
cursor = conn.cursor()
```

```python


curr_time = '2022-07-10 07:10:00'

# 1. 연도 계산 (3년 더하기)
year_plus_3 = str(int(curr_time[:4]) + 3)

# 2. 월-일 부분 (예: '-12-10 ')
month_day = curr_time[4:11]

# 3. '시' 부분만 추출해서 0 제거 (예: '03' -> 3 -> '3')
# 인덱스 11:13은 'HH' 부분을 의미합니다.
hour_int = int(curr_time[11:13])
hour_str = str(hour_int)

# 4. 최종 DB 포맷 조립 (예: '2025-12-10 3')
current_hour_raw = f"{year_plus_3}{month_day}{hour_str}"
print(current_hour_raw)


query = """
    SELECT * FROM weather_table 
    WHERE 일시 LIKE ?
"""

query2 = """
    SELECT 
        b.latitude, 
        b.longitude, 
        w."기온(°C)"
    FROM 
        weather_buoy AS w
    JOIN 
        buoy_position AS b ON w.지점 = b.지점
    WHERE 
        w.일시 LIKE ?
"""

# 3. 쿼리 실행
search_param = f"{current_hour_raw.strip()}:%"
print(search_param)
cursor.execute(query2, (search_param,))
rows = cursor.fetchall()

# 4. 데이터 출력 및 처리
cnt = 0
if rows:
    print("총 개수")
    
    print(f"🔔 [시간 변경 감지] {current_hour_raw}시 기상 데이터 갱신")
    for row in rows:
        print(row)
        cnt += 1

print(cnt)
    
```
